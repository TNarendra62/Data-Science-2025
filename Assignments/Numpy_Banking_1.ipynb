```python
# Question 8: Sorting and Searching
# Q8.1: Sort the banking_data by account balance (ascending)

# Q8.2: Sort the banking_data by credit score (descending)

# Q8.3: Find the indices that would sort the array by years active

# Q8.4: Find the customer with the second highest account balance

# Q8.5: Find the customer with the lowest transaction count
```

```python
sorted_by_balance = banking_data[banking_data[:, 1].argsort()]
print(sorted_by_balance)
```

```python
sorted_by_credit = banking_data[(-banking_data[:, 2]).argsort()]
print(sorted_by_credit)
```

```python
indices_yearsactive = banking_data[:, 4].argsort()
print(indices_yearsactive)
```

```python
sorted_balance_desc = banking_data[(-banking_data[:, 1]).argsort()]
second_highest = sorted_balance_desc[1]
print(second_highest)
```
